Emergency Incident Intake and Vehicle Lookup System using Priority Queues, Hash Tables, and String Processing

The Ridgeview Metropolitan Council currently manages emergency-vehicle dispatch (ambulance, fire, and traffic-response units) using a single spreadsheet-based log maintained by hand. All incoming incident reports – regardless of severity – are appended to the bottom of a single Python list by a call-centre application and are processed strictly in the order they were entered. The Council has approved funding for a pilot system, SafeRoute, and has hired you as a junior software developer/analyst. You have been given the following functional requirements, gathered from Council stakeholders:

 

R1 – Incident intake: Incoming calls must be logged as they arrive and processed in the order received, EXCEPT those life-threatening incidents (e.g. cardiac arrest, structure fire with trapped people) must be moved ahead of routine incidents (e.g. noise complaint, minor fender-bender) regardless of arrival order.
R2 – Vehicle and crew lookup: Dispatchers must be able to look up a specific emergency vehicle's status by its registration/fleet number almost instantly, even once the fleet grows to tens of thousands of vehicles across the wider provincial network.
R3 – Call transcript processing: Call-centre operators type a free-text transcript for every call. The system must be able to validate and extract structured information from this text (e.g. a vehicle registration plate mentioned by a witness, or a suburb name) before it can be used elsewhere in the system. The Council's existing developer has expressed concern that ‘a list is a list – we don't need anything more complicated than what we already have’. You have been asked to produce a short technical project report that responds to this position and delivers a working proof-of-concept for parts of the system.

Road Network Route-Finding and Suburb Reporting System using Graph Traversal, Recursive Algorithms, and Tree/Hash Indexing

Six months into the SafeRoute pilot described in Scenario 1, the Council has approved a second phase: connecting the intake and lookup systems from Phase 1 to a live model of Ridgeview's road network and building a proper reporting layer on top of the incident data the system has been collecting. R4 – Route-finding: The road network must be modelled so that the system can determine whether a path exists between the dispatch depot and an incident location and can traverse the network to identify candidate routes.
R5 – Historical audit trail: Auditors need to be able to browse the chronological history of incidents in a given suburb, and dispatch supervisors have asked for the ability to 'undo' the last dispatch action taken in error during a shift.
